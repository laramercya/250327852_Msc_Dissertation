```python
import cv2
import numpy as np
from skimage.morphology import skeletonize
import warnings

warnings.filterwarnings("ignore")

print("Image-processing libraries loaded.")
```

```python
CROP_TOP = 0.07
CROP_BOTTOM = 0.07
CROP_LEFT = 0.0
CROP_RIGHT = 0.0

VALUE_MIN = 40
PANEL_HUE_MIN = 100
PANEL_HUE_MAX = 135
PANEL_SAT_MIN = 50

MIN_COMP_AREA = 300

FEATURE_NAMES = [
    "F01_bead_pixel_ratio",
    "F02_num_components",
    "F03_largest_area_ratio",
    "F04_component_separation_flag",
    "F05_component_separation_distance_px",
    "F06_skeleton_length",
    "F07_skeleton_ratio",
    "F08_edge_count",
    "F09_edge_density",
    "F10_mean_hue",
    "F11_std_hue",
    "F12_mean_saturation",
    "F13_mean_value",
    "F14_hue_low_ratio",
    "F15_hue_high_ratio"
]

print("Image-processing parameters ready.")
print(f"Number of features: {len(FEATURE_NAMES)}")
```

```python
def preprocess(image):
    h, w = image.shape[:2]

    cropped = image[
        int(h * CROP_TOP):h - int(h * CROP_BOTTOM),
        int(w * CROP_LEFT):w - int(w * CROP_RIGHT)
    ]

    return cropped


def extract_foreground(hsv):
    h = hsv[:, :, 0]
    s = hsv[:, :, 1]
    v = hsv[:, :, 2]

    # Remove dark/background pixels
    valid_value = v >= VALUE_MIN

    # Identify the blue panel region
    panel_region = (
        (h >= PANEL_HUE_MIN) &
        (h <= PANEL_HUE_MAX) &
        (s >= PANEL_SAT_MIN)
    )

    # Remaining pixels form the foreground/sealant representation
    mask = (
        valid_value &
        ~panel_region
    ).astype(np.uint8) * 255

    return mask


print("Preprocessing and segmentation functions ready.")
```

```python
def connected_components(mask):
    n, _, stats, centroids = cv2.connectedComponentsWithStats(
        mask,
        connectivity=8
    )

    components = []

    for i in range(1, n):
        area = stats[i, cv2.CC_STAT_AREA]

        if area >= MIN_COMP_AREA:
            components.append({
                "area": area,
                "centroid": centroids[i]
            })

    # Largest component first
    components.sort(key=lambda c: -c["area"])

    return components


def skeletonise(mask):
    skeleton = skeletonize(mask > 0).astype(np.uint8) * 255
    skeleton_length = int(np.count_nonzero(skeleton))

    return skeleton, skeleton_length


def edge_analysis(gray):
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)

    edges = cv2.Canny(
        blurred,
        50,
        150
    )

    edge_count = int(np.count_nonzero(edges))

    edge_density = edge_count / max(edges.size, 1)

    return edges, edge_count, edge_density


print("Connected-component, skeletonisation and edge-analysis functions ready.")
```

```python
def extract_features(mask, hsv, comps, skel_len, ec, ed):
    bp = int(np.count_nonzero(mask))
    tp = mask.size

    f = {}

    # 1. Segmented region / structural features
    f["F01_bead_pixel_ratio"] = bp / max(tp, 1)

    f["F02_num_components"] = float(len(comps))

    f["F03_largest_area_ratio"] = (
        comps[0]["area"] / max(bp, 1)
        if comps else 0.0
    )

    # Component separation indicator
    f["F04_component_separation_flag"] = (
        1.0 if len(comps) >= 2 else 0.0
    )

    # Distance between the two largest retained components
    if len(comps) >= 2:
        c1 = comps[0]["centroid"]
        c2 = comps[1]["centroid"]

        f["F05_component_separation_distance_px"] = float(
            np.sqrt(
                (c1[0] - c2[0]) ** 2 +
                (c1[1] - c2[1]) ** 2
            )
        )
    else:
        f["F05_component_separation_distance_px"] = 0.0

    # 2. Skeleton features
    f["F06_skeleton_length"] = float(skel_len)

    f["F07_skeleton_ratio"] = (
        skel_len / max(bp, 1)
    )

    # 3. Edge features
    f["F08_edge_count"] = float(ec)

    f["F09_edge_density"] = ed

    # 4. HSV colour features
    bm = mask > 0

    hc = hsv[:, :, 0][bm].astype(np.float32)
    sc = hsv[:, :, 1][bm].astype(np.float32)
    vc = hsv[:, :, 2][bm].astype(np.float32)

    if len(hc) > 0:

        f["F10_mean_hue"] = float(np.mean(hc))
        f["F11_std_hue"] = float(np.std(hc))
        f["F12_mean_saturation"] = float(np.mean(sc))
        f["F13_mean_value"] = float(np.mean(vc))

        f["F14_hue_low_ratio"] = float(
            np.mean(hc < 30)
        )

        f["F15_hue_high_ratio"] = float(
            np.mean(hc > 90)
        )

    else:

        for k in [
            "F10_mean_hue",
            "F11_std_hue",
            "F12_mean_saturation",
            "F13_mean_value",
            "F14_hue_low_ratio",
            "F15_hue_high_ratio"
        ]:
            f[k] = 0.0

    return f


def extract(image):

    crop = preprocess(image)

    hsv = cv2.cvtColor(
        crop,
        cv2.COLOR_BGR2HSV
    )

    gray = cv2.cvtColor(
        crop,
        cv2.COLOR_BGR2GRAY
    )

    mask = extract_foreground(hsv)

    comps = connected_components(mask)

    skel, skeleton_length = skeletonise(mask)

    edges, edge_count, edge_density = edge_analysis(gray)

    features = extract_features(
        mask,
        hsv,
        comps,
        skeleton_length,
        edge_count,
        edge_density
    )

    intermediates = {
        "crop": crop,
        "mask": mask,
        "skel": skel,
        "edges": edges,
        "components": comps
    }

    return features, intermediates


def feature_vector(features):
    return [
        features[k]
        for k in FEATURE_NAMES
    ]


print("Feature extraction pipeline ready.")
print(f"Features per image: {len(FEATURE_NAMES)}")
```

```python
import time

IMAGE_DIR = Path("images")
all_paths = sorted(IMAGE_DIR.glob("*.bmp"))

print(f"Found {len(all_paths)} BMP images")

X = []
y = []
filenames = []
intermediates_list = []
processing_times = []

for path in all_paths:

    img = cv2.imread(str(path))

    if img is None:
        print(f"SKIP: {path.name}")
        continue

    t0 = time.perf_counter()

    # Complete image-processing pipeline
    features, intermediates = extract(img)

    processing_time_ms = (
        time.perf_counter() - t0
    ) * 1000

    label = LABELS.get(path.name, "")

    X.append(feature_vector(features))
    y.append(label)
    filenames.append(path.name)
    intermediates_list.append(intermediates)
    processing_times.append(processing_time_ms)

    print(
        f"{path.name:<12} "
        f"label={label:<5} "
        f"time={processing_time_ms:.2f} ms"
    )

X = np.array(X, dtype=float)
y = np.array(y)

processing_times = np.array(processing_times)

# Check that every image has a valid label
valid_labels = np.isin(y, ["good", "fail"])

print("\n" + "=" * 55)
print("FEATURE EXTRACTION SUMMARY")
print("=" * 55)

print(f"Images processed : {len(y)}")
print(f"Good images      : {np.sum(y == 'good')}")
print(f"Fail images      : {np.sum(y == 'fail')}")
print(f"Feature matrix   : {X.shape}")
print(f"Valid labels     : {np.sum(valid_labels)}/{len(y)}")

print("\nProcessing time:")
print(f"Mean   : {np.mean(processing_times):.2f} ms")
print(f"Median : {np.median(processing_times):.2f} ms")
print(f"Maximum: {np.max(processing_times):.2f} ms")
print("=" * 55)
```

```python
import pandas as pd

feature_df = pd.DataFrame(
    X,
    columns=FEATURE_NAMES
)

feature_df.insert(0, "filename", filenames)
feature_df.insert(1, "label", y)

print("Feature dataset shape:", feature_df.shape)

display(feature_df.head(10))
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

# Feature matrix and labels
X_ml = feature_df[FEATURE_NAMES].values
y_ml = feature_df["label"].values

# Convert labels to binary values
y_binary = np.where(y_ml == "fail", 1, 0)

print("X shape:", X_ml.shape)
print("y shape:", y_binary.shape)

print("\nClass distribution:")
print("Good:", np.sum(y_binary == 0))
print("Fail:", np.sum(y_binary == 1))

# Five-fold stratified cross-validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

print("\n5-fold stratified cross-validation ready.")
```

```python
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
import numpy as np
import pandas as pd

models = {
    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        class_weight="balanced",
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),

    "SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", SVC(
            kernel="rbf",
            class_weight="balanced",
            probability=True,
            random_state=42
        ))
    ])
}

results = []
predictions = {}

for model_name, model in models.items():

    all_true = []
    all_pred = []

    for train_idx, test_idx in cv.split(X_ml, y_binary):

        X_train, X_test = X_ml[train_idx], X_ml[test_idx]
        y_train, y_test = y_binary[train_idx], y_binary[test_idx]

        model.fit(X_train, y_train)

        y_pred = model.predict(X_test)

        all_true.extend(y_test)
        all_pred.extend(y_pred)

    all_true = np.array(all_true)
    all_pred = np.array(all_pred)

    accuracy = accuracy_score(all_true, all_pred)
    precision = precision_score(
        all_true, all_pred, pos_label=1, zero_division=0
    )
    recall = recall_score(
        all_true, all_pred, pos_label=1, zero_division=0
    )
    f1 = f1_score(
        all_true, all_pred, pos_label=1, zero_division=0
    )
    macro_f1 = f1_score(
        all_true, all_pred, average="macro", zero_division=0
    )

    cm = confusion_matrix(all_true, all_pred)

    results.append({
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "Macro-F1": macro_f1
    })

    predictions[model_name] = {
        "true": all_true,
        "pred": all_pred,
        "confusion_matrix": cm
    }

results_df = pd.DataFrame(results)

print("=" * 70)
print("5-FOLD STRATIFIED CROSS-VALIDATION RESULTS")
print("=" * 70)

display(
    results_df.style.format({
        "Accuracy": "{:.3f}",
        "Precision": "{:.3f}",
        "Recall": "{:.3f}",
        "F1": "{:.3f}",
        "Macro-F1": "{:.3f}"
    })
)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

for model_name, data in predictions.items():

    cm = data["confusion_matrix"]

    print(f"\n{model_name}")
    print(cm)

    disp = ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Good", "Fail"]
    )

    disp.plot()
    plt.title(f"{model_name} - 5-Fold Cross-Validation")
    plt.show()
```

```python
# ============================================================
# METHOD 1 — 100 IMAGE OUT-OF-FOLD RESULTS
# Gradient Boosting + 5-Fold Stratified CV
# ============================================================

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score
import numpy as np
import os
import cv2
import matplotlib.pyplot as plt



oof_pred = np.zeros(len(X_ml), dtype=int)
oof_prob = np.zeros(len(X_ml), dtype=float)
oof_fold = np.zeros(len(X_ml), dtype=int)

gb_model = GradientBoostingClassifier(
    random_state=42
)



for fold, (train_idx, test_idx) in enumerate(cv.split(X_ml, y_binary), start=1):

    X_train = X_ml[train_idx]
    X_test = X_ml[test_idx]

    y_train = y_binary[train_idx]

    gb_model.fit(X_train, y_train)

    # Prediction
    pred = gb_model.predict(X_test)

    # Probability of Fail
    prob = gb_model.predict_proba(X_test)[:, 1]

    oof_pred[test_idx] = pred
    oof_prob[test_idx] = prob
    oof_fold[test_idx] = fold



oof_accuracy = accuracy_score(y_binary, oof_pred)

print("=" * 60)
print("METHOD 1 — OUT-OF-FOLD RESULTS")
print("=" * 60)

print(f"Images       : {len(y_binary)}")
print(f"Good         : {np.sum(y_binary == 0)}")
print(f"Fail         : {np.sum(y_binary == 1)}")
print(f"OOF Accuracy : {oof_accuracy:.3f} ({oof_accuracy*100:.1f}%)")



method1_results = feature_df[["filename", "label"]].copy()

method1_results["true_numeric"] = y_binary
method1_results["prediction_numeric"] = oof_pred
method1_results["prediction"] = np.where(
    oof_pred == 1, "fail", "good"
)

method1_results["fail_probability"] = oof_prob
method1_results["confidence"] = np.maximum(
    oof_prob,
    1 - oof_prob
)

method1_results["correct"] = (
    method1_results["true_numeric"]
    == method1_results["prediction_numeric"]
)

method1_results["fold"] = oof_fold

print("\nResults created successfully.")
display(method1_results.head(10))
```

```python
# ============================================================
# METHOD 1 — CREATE 4 VISUAL RESULT SHEETS
# ============================================================

import os
import cv2
import matplotlib.pyplot as plt

OUTPUT_DIR = "/content/method1_result_sheets"
os.makedirs(OUTPUT_DIR, exist_ok=True)



for sheet_number in range(4):

    start = sheet_number * 25
    end = min(start + 25, len(method1_results))

    subset = method1_results.iloc[start:end]

    fig, axes = plt.subplots(
        5,
        5,
        figsize=(20, 20)
    )

    axes = axes.flatten()

    for ax, (_, row) in zip(axes, subset.iterrows()):

        filename = row["filename"]

        image_path = os.path.join(
            "/content/images",
            filename
        )

        image = cv2.imread(image_path)

        if image is not None:

            image = cv2.cvtColor(
                image,
                cv2.COLOR_BGR2RGB
            )

            ax.imshow(image)


        # Results


        true_label = str(row["label"]).upper()
        pred_label = str(row["prediction"]).upper()

        confidence = float(row["confidence"]) * 100

        if bool(row["correct"]):
            symbol = "✓"
        else:
            symbol = "✗"

        ax.set_title(
            f"{filename}\n"
            f"TRUE: {true_label} | PRED: {pred_label}\n"
            f"{confidence:.1f}% {symbol}",
            fontsize=9
        )

        ax.axis("off")

    # Hide unused axes
    for ax in axes[len(subset):]:
        ax.axis("off")



    fig.suptitle(
        f"Sealant Bead Inspection — Method 1 Results "
        f"({start+1}–{end})",
        fontsize=18,
        fontweight="bold"
    )

    plt.tight_layout()


    output_path = os.path.join(
        OUTPUT_DIR,
        f"method1_results_{start+1}_{end}.png"
    )

    plt.savefig(
        output_path,
        dpi=200,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

    print(f"Saved: {output_path}")

print("\n==========================================")
print("ALL 4 RESULT SHEETS CREATED")
print("==========================================")
print(f"Location: {OUTPUT_DIR}")
```

```python
# ============================================================
# METHOD 1 — GRADIENT BOOSTING
# 5-FOLD OUT-OF-FOLD EVALUATION + CONFUSION MATRIX
# ============================================================

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import matplotlib.pyplot as plt



feature_columns = [
    col for col in feature_df.columns
    if col.startswith("F")
]

print("Number of features:", len(feature_columns))



X = feature_df[feature_columns]

y = feature_df["label"].map({
    "good": 0,
    "fail": 1
})

print("X shape:", X.shape)
print("y shape:", y.shape)



cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


model = GradientBoostingClassifier(
    random_state=42
)

# Generate out-of-fold predictions
y_pred = cross_val_predict(
    model,
    X,
    y,
    cv=cv
)



accuracy = accuracy_score(y, y_pred)

precision = precision_score(
    y,
    y_pred,
    average="macro"
)

recall = recall_score(
    y,
    y_pred,
    average="macro"
)

macro_f1 = f1_score(
    y,
    y_pred,
    average="macro"
)

print("\n" + "=" * 60)
print("GRADIENT BOOSTING — 5-FOLD OUT-OF-FOLD RESULTS")
print("=" * 60)

print(f"Accuracy  : {accuracy:.3f}")
print(f"Precision : {precision:.3f}")
print(f"Recall    : {recall:.3f}")
print(f"Macro-F1  : {macro_f1:.3f}")



print("\nCLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y,
        y_pred,
        target_names=["Good", "Fail"],
        digits=3
    )
)


# Confusion matrix

cm = confusion_matrix(y, y_pred)

print("\nCONFUSION MATRIX")
print("=" * 60)
print(cm)

# Display confusion matrix

fig, ax = plt.subplots(figsize=(7, 6))

im = ax.imshow(cm)

ax.set_xticks([0, 1])
ax.set_yticks([0, 1])

ax.set_xticklabels(["Good", "Fail"])
ax.set_yticklabels(["Good", "Fail"])

ax.set_xlabel("Predicted Class")
ax.set_ylabel("True Class")

ax.set_title(
    "Gradient Boosting — 5-Fold Cross-Validation"
)

for i in range(2):
    for j in range(2):
        ax.text(
            j,
            i,
            cm[i, j],
            ha="center",
            va="center",
            fontsize=16
        )

plt.colorbar(im)

plt.tight_layout()

output_path = "/content/gradient_boosting_confusion_matrix.png"

plt.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\nSaved:", output_path)
```

```python
# METHOD 1 — RANDOM FOREST

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np



feature_columns = [
    "F01_bead_pixel_ratio",
    "F02_num_components",
    "F03_largest_area_ratio",
    "F04_component_separation_flag",
    "F05_component_separation_distance_px",
    "F06_skeleton_length",
    "F07_skeleton_ratio",
    "F08_edge_count",
    "F09_edge_density",
    "F10_mean_hue",
    "F11_std_hue",
    "F12_mean_saturation",
    "F13_mean_value",
    "F14_hue_low_ratio",
    "F15_hue_high_ratio"
]

X = feature_df[feature_columns]
y = feature_df["label"]

print("Number of features:", X.shape[1])
print("X shape:", X.shape)
print("y shape:", y.shape)



cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)



rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

# Out-of-fold predictions
rf_pred = cross_val_predict(
    rf_model,
    X,
    y,
    cv=cv,
    method="predict"
)

# ------------------------------------------------------------
# Metrics
# ------------------------------------------------------------

accuracy = accuracy_score(y, rf_pred)

precision = precision_score(
    y,
    rf_pred,
    pos_label="fail"
)

recall = recall_score(
    y,
    rf_pred,
    pos_label="fail"
)

macro_f1 = f1_score(
    y,
    rf_pred,
    average="macro"
)

print("\n" + "=" * 60)
print("RANDOM FOREST — 5-FOLD OUT-OF-FOLD RESULTS")
print("=" * 60)

print(f"Accuracy  : {accuracy:.3f}")
print(f"Precision : {precision:.3f}")
print(f"Recall    : {recall:.3f}")
print(f"Macro-F1  : {macro_f1:.3f}")


# Classification report

print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y,
        rf_pred,
        labels=["good", "fail"],
        target_names=["Good", "Fail"],
        digits=3
    )
)

# Confusion matrix

cm = confusion_matrix(
    y,
    rf_pred,
    labels=["good", "fail"]
)

print("\n" + "=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print(cm)


# Plot confusion matrix


plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="viridis",
    xticklabels=["Good", "Fail"],
    yticklabels=["Good", "Fail"]
)

plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.title("Random Forest — 5-Fold Cross-Validation")

plt.tight_layout()

output_path = "/content/random_forest_confusion_matrix.png"

plt.savefig(
    output_path,
    dpi=200,
    bbox_inches="tight"
)

plt.show()

print("\nSaved:", output_path)
```

```python

# METHOD 1 — SVM


from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)



X = feature_df[feature_columns]
y = feature_df["label"]

print("Number of features:", X.shape[1])
print("X shape:", X.shape)
print("y shape:", y.shape)


cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)



svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale"
    ))
])



svm_pred = cross_val_predict(
    svm_model,
    X,
    y,
    cv=cv,
    method="predict"
)



accuracy = accuracy_score(y, svm_pred)

precision = precision_score(
    y,
    svm_pred,
    pos_label="fail"
)

recall = recall_score(
    y,
    svm_pred,
    pos_label="fail"
)

macro_f1 = f1_score(
    y,
    svm_pred,
    average="macro"
)

print("\n" + "=" * 60)
print("SVM — 5-FOLD OUT-OF-FOLD RESULTS")
print("=" * 60)

print(f"Accuracy  : {accuracy:.3f}")
print(f"Precision : {precision:.3f}")
print(f"Recall    : {recall:.3f}")
print(f"Macro-F1  : {macro_f1:.3f}")



print("\n" + "=" * 60)
print("CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y,
        svm_pred,
        labels=["good", "fail"],
        target_names=["Good", "Fail"],
        digits=3
    )
)



cm = confusion_matrix(
    y,
    svm_pred,
    labels=["good", "fail"]
)

print("\n" + "=" * 60)
print("CONFUSION MATRIX")
print("=" * 60)

print(cm)



plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="viridis",
    xticklabels=["Good", "Fail"],
    yticklabels=["Good", "Fail"]
)

plt.xlabel("Predicted Class")
plt.ylabel("True Class")
plt.title("SVM — 5-Fold Cross-Validation")

plt.tight_layout()

output_path = "/content/svm_confusion_matrix.png"

plt.savefig(
    output_path,
    dpi=200,
    bbox_inches="tight"
)

plt.show()

print("\nSaved:", output_path)
```

```python
# ============================================================
# METHOD 1 — COLLECT ALL OUTPUTS INTO ONE ZIP FILE
# ============================================================

import os
import glob
import zipfile

OUTPUT_ZIP = "/content/METHOD_1_ALL_OUTPUTS.zip"

# Files/folders we want to include
patterns = [
    "/content/method1_model_comparison.csv",
    "/content/method1_model_comparison.png",

    "/content/gradient_boosting_confusion_matrix.png",
    "/content/random_forest_confusion_matrix.png",
    "/content/svm_confusion_matrix.png",

    "/content/random_forest_feature_importance.png",

    "/content/method1_result_sheets/*"
]

files_to_zip = []

for pattern in patterns:
    files_to_zip.extend(glob.glob(pattern))

# Remove duplicates
files_to_zip = sorted(list(set(files_to_zip)))

print("Files found:")
for f in files_to_zip:
    print("✓", f)

# Create ZIP
with zipfile.ZipFile(OUTPUT_ZIP, "w", zipfile.ZIP_DEFLATED) as zipf:

    for file_path in files_to_zip:

        if os.path.isfile(file_path):
            zipf.write(
                file_path,
                arcname=os.path.basename(file_path)
            )

print("\n" + "=" * 60)
print("METHOD 1 ZIP CREATED")
print("=" * 60)
print("Saved:", OUTPUT_ZIP)
print("Number of files:", len(files_to_zip))
```
